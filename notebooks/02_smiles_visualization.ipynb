# 02 — European index implied-volatility smiles

Compare the observed (synthetic) IV curves for several expiries. The horizontal coordinate is log-forward-moneyness $k=\log(K/F_T)$ and each observation comes from inverting a synthetic option mid price.

```python
import matplotlib.pyplot as plt
from volsurface.demo import synthetic_chain
from volsurface.cleaning import prepare_options
raw=synthetic_chain()
df=prepare_options(raw)
fig,ax=plt.subplots(figsize=(12,7))
for expiry,g in df.groupby("expiry_date"):
    g=g.sort_values("log_moneyness")
    ax.plot(g.log_moneyness, 100*g.iv, marker=".", label=expiry)
ax.set(xlabel="Log-forward-moneyness",ylabel="Implied volatility (%)",title="Synthetic European index option volatility smiles")
ax.grid(alpha=.3)
ax.legend(ncol=2)
plt.show()
```

## Interpretation

A left-wing put skew means lower strikes often command higher implied volatility than near-the-money strikes. Here the profile is designed for illustration; it must not be interpreted as an empirical estimate of today’s EURO STOXX 50 skew.
