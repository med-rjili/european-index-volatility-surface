# 01 — Synthetic EURO STOXX 50-like options and Black–76 inversion

All sample quotes are synthetic, not Eurex market observations. This notebook shows Black–76 prices, implied-volatility inversion with Brent’s method, and data-quality filtering.

```python
import numpy as np
from volsurface.demo import synthetic_chain
from volsurface.cleaning import black76_price, implied_volatility, prepare_options
raw = synthetic_chain(seed=2026)
raw.head()
```

```python
F, K, T, sigma = 5000, 4800, 0.25, 0.20
price = black76_price(F, K, T, sigma, option_type="put", discount=np.exp(-0.02*T))
recovered = implied_volatility(price,F,K,T,option_type="put",discount=np.exp(-0.02*T))
print(f"Synthetic Black–76 put: {price:.3f} index points; recovered IV: {recovered:.6f}")
```

```python
clean = prepare_options(raw)
print("Raw",len(raw),"Clean",len(clean))
clean[["expiry_date","strike","forward","mid","iv","log_moneyness","total_variance"]].head()
```
